```python
import chess
import chess.engine
import random
from collections import defaultdict

# Change these paths to your engines
engine1_path = "/Users/camiloariasabad/Downloads/stockfish/stockfish-macos-m1-apple-silicon"  # e.g., Stockfish
engine2_path = "/Users/camiloariasabad/Downloads/stockfish/stockfish-macos-m1-apple-silicon"  # e.g., a student bot

games_to_play = 10
time_limit = 0.001  # seconds per move
results = defaultdict(int)  # '1-0', '0-1', '1/2-1/2'

def play_one_game(white_path, black_path):
    board = chess.Board()
    with chess.engine.SimpleEngine.popen_uci(white_path) as white_engine, \
         chess.engine.SimpleEngine.popen_uci(black_path) as black_engine:

        engines = [white_engine, black_engine]
        turn = 0
        while not board.is_game_over():
            move = engines[turn].play(board, chess.engine.Limit(time=time_limit)).move
            board.push(move)
            turn ^= 1
        return board.result()

# Play games
for i in range(games_to_play):
    if i % 2 == 0:
        result = play_one_game(engine1_path, engine2_path)  # engine1 as white
    else:
        result = play_one_game(engine2_path, engine1_path)  # engine2 as white
        # Flip result
        if result == "1-0":
            result = "0-1"
        elif result == "0-1":
            result = "1-0"

    results[result] += 1
    if (i + 1) % 100 == 0:
        print(f"{i + 1} games played...")

# Print final results
print("\nFinal Results after 1000 games:")
print("Engine1 wins:", results["1-0"])
print("Engine2 wins:", results["0-1"])
print("Draws:", results["1/2-1/2"])
```
