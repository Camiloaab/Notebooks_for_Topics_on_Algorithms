```python
import networkx as nx
import matplotlib.pyplot as plt

# Create graph
G = nx.Graph()
nodes = list(range(5, 15))
G.add_nodes_from(nodes)

# Add edges based on i + j divisible by 5
for i in nodes:
    for j in nodes:
        if i < j and (i + j) % 5 == 0:
            G.add_edge(i, j)

# Draw with circular layout
pos = nx.circular_layout(G)
nx.draw(G, pos, with_labels=True, node_color="lightblue", edge_color="gray", node_size=500)
plt.title("Grafo")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
from itertools import combinations

# ───────────────────────────────────────────────────────────────
# 1.  Tabla de habilidades  (X = la persona posee la habilidad)
# ───────────────────────────────────────────────────────────────
habilidades = {
    "Jugar ajedrez":       {"Ana", "Carla","Elena"},                # fila 1
    "Hablar japonés":      {"Ana", "Beto","David","Elena"},                 # fila 2
    "Nadar":               {"Beto", "Carla", "David"},      # fila 3
    "Montar en bicicleta": {"Ana", "Beto","David", "Elena"},       # fila 4
    "Cocinar":             {"Beto", "David", "Elena"},      # fila 5
}

personas = ["Ana", "Beto", "Carla", "David", "Elena"]
umbral   = 2        # mínimo de habilidades compartidas para unir

# ───────────────────────────────────────────────────────────────
# 2.  Crear grafo y añadir aristas según el umbral
# ───────────────────────────────────────────────────────────────
G = nx.Graph()
G.add_nodes_from(personas)

# Para cada pareja de personas, contar cuántas habilidades comparten
for p, q in combinations(personas, 2):
    compartidas = sum({p, q}.issubset(grupo) for grupo in habilidades.values())
    if compartidas >= umbral:
        G.add_edge(p, q, peso=compartidas)   # opcional: guardar # de coincidencias

# ───────────────────────────────────────────────────────────────
# 3.  Dibujar con disposición circular
# ───────────────────────────────────────────────────────────────
pos = nx.circular_layout(G)
edge_labels = {(u, v): d["peso"] for u, v, d in G.edges(data=True)}

nx.draw_networkx_nodes(G, pos, node_color="lightblue", node_size=1500)
nx.draw_networkx_labels(G, pos, font_size=11)
nx.draw_networkx_edges(G, pos, edge_color="gray", width=1.5)
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_color="red")

plt.title("Arista si comparten ≥2 habilidades")
plt.axis("off")
plt.show()

# ───────────────────────────────────────────────────────────────
# 4.  Resultado por consola
# ───────────────────────────────────────────────────────────────
print("Aristas (peso = nº de habilidades compartidas):")
for u, v, d in G.edges(data=True):
    print(f"{u:5s} – {v:5s}   ({d['peso']})")
```

```python
import networkx as nx

# ───────────────────────────────────────────────────────────
# 1. Definir el grafo ponderado según la tabla
# ───────────────────────────────────────────────────────────
G = nx.Graph()

# Lista de aristas (u, v, peso) tomadas de la matriz
aristas = [
    ("A", "B", 3), ("A", "D", 2),
    ("B", "C", 4), ("B", "E", 1),
    ("C", "D", 5), ("C", "F", 7),
    ("D", "E", 6),
    ("E", "F", 2), ("E", "G", 3),
    ("F", "G", 4),
]

G.add_weighted_edges_from(aristas)   # añade aristas con atributo 'weight'

# ───────────────────────────────────────────────────────────
# 2. Calcular la distancia mínima de A a E
# ───────────────────────────────────────────────────────────
d = nx.dijkstra_path_length(G, source="A", target="E", weight="weight")
ruta = nx.dijkstra_path(G, source="A", target="E", weight="weight")

print(f"Distancia mínima d(A,E) = {d}")
print("Ruta más corta:", " → ".join(ruta))
```
